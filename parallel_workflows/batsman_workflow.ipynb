```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict

```

```python
class BatsmanState(TypedDict):
    runs:int
    balls:int
    fours:int
    sixes:int

    strike_rate:float
    boundary_per_ball:float
    boundary_percentage:float
```

```python
def calculate_sr(state:BatsmanState):
    sr=(state['runs']/state['balls'])*100
    return {'strike_rate':sr}

def calculate_bpb(state:BatsmanState):
    bp=(state['fours']+state['sixes'])/state['balls']
    return {'boundary_per_ball':bp}

def calculate_bp(state:BatsmanState):
    bp=((state['fours']*4)+(state['sixes']*6))/state['runs']*100
    return {'boundary_percentage':bp}

def summary(state:BatsmanState):
    print("Runs: ",state['runs'])
    print("Balls: ",state['balls'])
    print("Fours: ",state['fours'])
    print("Sixes: ",state['sixes'])
    print("Strike Rate: ",state['strike_rate'])
    print("Boundary per Ball: ",state['boundary_per_ball'])
    print("Boundary Percentage: ",state['boundary_percentage'])
    return state
```

```python
graph=StateGraph(BatsmanState)
graph.add_node('calculate_strike_rate',calculate_sr)
graph.add_node('calculate_bpb',calculate_bpb)
graph.add_node('calculate_bp',calculate_bp)
graph.add_node('summary',summary)
```

```python
graph.add_edge(START,'calculate_strike_rate')
graph.add_edge(START,'calculate_bpb')
graph.add_edge(START,'calculate_bp')
graph.add_edge('calculate_strike_rate','summary')
graph.add_edge('calculate_bpb','summary')
graph.add_edge('calculate_bp','summary')
graph.add_edge('summary',END)
```

```python
workflow=graph.compile()

initial_state=BatsmanState(runs=100,balls=60,fours=10,sixes=5)
result=workflow.invoke(initial_state)
```
