```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
#define state for BMI calculator
class BMIState(TypedDict):
    weight: float
    height: float
    bmi: float
    category: str
```

```python
def calculate_bmi(state:BMIState)->BMIState:
    weight = state['weight']
    height = state['height']
    bmi = weight / (height ** 2)
    state['bmi'] = bmi
    
    return state
def label_bmi_category(state:BMIState)->BMIState:
    bmi = state['bmi']
    if bmi < 18.5:
        state['category'] = 'Underweight'
    elif 18.5 <= bmi < 24.9:
        state['category'] = 'Normal weight'
    elif 25 <= bmi < 29.9:
        state['category'] = 'Overweight'
    else:
        state['category'] = 'Obesity'
    
    return state
```

```python
#define graph
graph=StateGraph(BMIState)

#add nodes to graph
graph.add_node("calculate_bmi",calculate_bmi)
graph.add_node("label_bmi_category",label_bmi_category)

#add edges to graph
graph.add_edge(START,"calculate_bmi")
graph.add_edge("calculate_bmi","label_bmi_category")
graph.add_edge("label_bmi_category",END)

#compile the graph
workflow=graph.compile()

#execute
result=workflow.invoke({"weight": 70, "height": 1.75})

print(result)
```
